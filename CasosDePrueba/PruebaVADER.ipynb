```python
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from nltk.tokenize import word_tokenize
import string

# Descarga de recursos (solo la primera vez)
nltk.download('punkt')
nltk.download('vader_lexicon')

# Inicializar VADER
sia = SentimentIntensityAnalyzer()

# Texto de ejemplo (positivo)
texto = "I love this game!!"

# Análisis de sentimiento SIN tokenizar
scores_sin_tokenizar = sia.polarity_scores(texto)

# Tokenización y limpieza: eliminar puntuación
tokens = word_tokenize(texto)
tokens_limpios = [t for t in tokens if t not in string.punctuation]
texto_reconstruido = " ".join(tokens_limpios)

# Análisis de sentimiento CON tokenización
scores_tokenizado = sia.polarity_scores(texto_reconstruido)

# Mostrar ambos resultados
import pandas as pd
df = pd.DataFrame({
    'neg': [scores_sin_tokenizar['neg'], scores_tokenizado['neg']],
    'neu': [scores_sin_tokenizar['neu'], scores_tokenizado['neu']],
    'pos': [scores_sin_tokenizar['pos'], scores_tokenizado['pos']],
    'compound': [scores_sin_tokenizar['compound'], scores_tokenizado['compound']],
}, index=['Sin tokenizar', 'Tokenizado'])

print(df)
```
