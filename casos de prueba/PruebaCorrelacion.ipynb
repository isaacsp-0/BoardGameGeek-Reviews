```python
import pandas as pd
import numpy as np

def test_corrwith():
    # Creamos una matriz de ejemplo: usuarios en filas, juegos en columnas
    # Usuario A y B puntúan Game1 y Game2 de la misma forma => correlación +1
    # Usuario C puntúa Game3 de forma opuesta a Game1 => correlación -1
    data = {
        'Game1': [5, 3, 1],
        'Game2': [5, 3, 1],   # idéntico a Game1
        'Game3': [1, 3, 5],   # opuesto a Game1
        'Game4': [5, 5, 5],   # sin variación => correlación NaN
    }
    games_matrix = pd.DataFrame(data, index=['A', 'B', 'C'])

    # Calculamos correlaciones de cada juego con Game1
    corr = games_matrix.corrwith(games_matrix['Game1'])

    # Redondeamos para facilitar la comparación
    corr = corr.round(2)

    # Mostramos los resultados
    print("Correlaciones con Game1:\n", corr, "\n")

    # Comprobaciones:
    assert corr['Game1'] == 1.00, "Autocorrelación de Game1 debe ser 1.0"
    assert corr['Game2'] == 1.00, "Game2 es idéntico a Game1, la correlación debe ser 1.0"
    assert corr['Game3'] == -1.00, "Game3 es inverso a Game1, la correlación debe ser -1.0"
    assert np.isnan(corr['Game4']), "Game4 no varía, correlación debe ser NaN"


# Ejecutamos el test
test_corrwith()
```
