```python
import numpy as np
import pandas as pd
from scipy.stats import zscore
```

```python
def arreglar_dispersion(df: pd.DataFrame, columnas, z_score_umbral: int):
    for col in columnas:
        z_scores = np.abs(zscore(df[col], nan_policy='omit'))  
        print(z_scores)
        df.loc[z_scores > z_score_umbral, col] = np.nan
```

```python
df_test = pd.DataFrame({'x': [0,1,2,3,1000]})
arreglar_dispersion(df_test, ['x'], z_score_umbral=1)
df_test.head()
```
