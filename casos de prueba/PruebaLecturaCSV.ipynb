```python
import pandas as pd
import random
```

```python
RATING_COLS = ['user',
               'rating',
               'ID', 
               'comment']

PROBABILIDAD_FALLO = 0.2
```

```python
reviews_completa = pd.read_csv("../data/bgg-26m-reviews.csv") 
reviews_completa.info(memory_usage='deep')
```

```python
reviews_optimizada = pd.read_csv("../data/bgg-26m-reviews.csv", 
                      usecols=RATING_COLS, 
                      skiprows=lambda x: x > 0 and random.random() >= PROBABILIDAD_FALLO,
                      dtype={'user': 'string', 
                             'rating': 'float16', 
                             'ID': 'uint32',
                             'comment': 'string[pyarrow]' #  Al usar string[pyarrow] se reduce la memoria significativamente
                             })
reviews_optimizada.info(memory_usage='deep')
```
